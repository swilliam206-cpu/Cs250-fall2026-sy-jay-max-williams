## Section 3 · Instance Methods
**About 5 minutes · 15 points**

A **method** is a function defined inside a class. It uses `self` to read and change the object’s own attributes.

**Your task:** complete `WaterBottle`.
- `WaterBottle(capacity)` stores `capacity`; the bottle starts empty (`amount = 0`).
- `fill(ml)` adds `ml`, but `amount` never goes above `capacity`.
- `drink(ml)` removes `ml`, but `amount` never goes below `0`. It **returns** how much was actually drunk.
- `is_empty()` returns `True` when `amount` is `0`.

```python

class WaterBottle:

    def __init__(self, capacity):
        # Store the maximum capacity of the bottle
        self.capacity = capacity

        # A new bottle starts empty
        self.amount = 0

    def fill(self, ml):
        # Add water without going above capacity
        self.amount = min(self.amount + ml, self.capacity)

    def drink(self, ml):
        # Calculate how much water can actually be drunk
        amount_drunk = min(ml, self.amount)

        # Remove the amount of water that was drunk
        self.amount = self.amount - amount_drunk

        # Return how much water was actually drunk
        return amount_drunk

    def is_empty(self):
        # Return True if the bottle has no water
        return self.amount == 0


# --------------------------------
# TEST THE CODE
# --------------------------------

# Create a new bottle with 1000 ml capacity
bottle = WaterBottle(1000)

# Check if the new bottle is empty
print("Is empty:", bottle.is_empty())

# Fill the bottle with 500 ml
bottle.fill(500)
print("Amount:", bottle.amount)

# Fill another 700 ml (should stop at 1000)
bottle.fill(700)
print("Amount after filling:", bottle.amount)

# Drink 300 ml
print("Amount drunk:", bottle.drink(300))
print("Amount remaining:", bottle.amount)

# Try drinking more water than is available
print("Amount drunk:", bottle.drink(900))
print("Amount remaining:", bottle.amount)

# Check if the bottle is empty
print("Is empty:", bottle.is_empty())

```

## Section 4 · Class & Static Methods
**About 4 minutes · 15 points**

- A **static method** (`@staticmethod`) gets no `self`: it is a helper that belongs with the class.
- A **class method** (`@classmethod`) gets the class as `cls`: it is often another way to build an object.

**Your task:** complete `Color`.
- `Color.is_valid(value)` returns `True` if `value` is an `int` from 0 to 255. Make it a **static method**.
- `Color.from_hex("#FF8800")` returns `Color(255, 136, 0)`. Make it a **class method**. Hint: `int("FF", 16)` is `255`.
- `Color(r, g, b)` raises `ValueError` if any channel is not valid (use `Color.is_valid`).

```python

# Create a class called Color
class Color:

    # Initialize the Color object with red, green, and blue values
    def __init__(self, r, g, b):

        # Check if all three color values are valid
        # Use the is_valid() method to check each value
        if not (Color.is_valid(r) and Color.is_valid(g) and Color.is_valid(b)):

            # Raise an error if any value is invalid
            raise ValueError("Color values must be integers from 0 to 255")

        # Store the red, green, and blue values
        self.r = r
        self.g = g
        self.b = b

    # A static method does not need self or cls
    # It checks whether a value is a valid color number
    @staticmethod
    def is_valid(value):

        # Check that the value is an integer from 0 to 255
        # type() is used instead of isinstance() to reject True and False
        return type(value) == int and 0 <= value <= 255

    # A class method works with the class itself using cls
    @classmethod
    def from_hex(cls, hex_string):

        # Remove the # and convert each pair of characters
        # from hexadecimal to a regular integer
        r = int(hex_string[1:3], 16)
        g = int(hex_string[3:5], 16)
        b = int(hex_string[5:7], 16)

        # Create and return a new Color object
        return cls(r, g, b)


# ---------------------------------
# TEST THE COLOR CLASS
# ---------------------------------

# Create a color using RGB values
color1 = Color(255, 136, 0)

# Print the RGB values
print("Red:", color1.r)
print("Green:", color1.g)
print("Blue:", color1.b)

# Check if a number is a valid color value
print("Is 255 valid?", Color.is_valid(255))
print("Is 300 valid?", Color.is_valid(300))

# Create a color using a hexadecimal value
color2 = Color.from_hex("#FF8800")

# Print the converted RGB values
print("Hexadecimal color:", color2.r, color2.g, color2.b)
```

## Section 5 · Dunder Methods
**About 5 minutes · 15 points**

Dunder (double underscore) methods let your objects work with Python’s built-in syntax. Python calls them for you.

**Your task:** add three dunder methods to `Movie`.
- `__str__`: `print(Movie("Inception", 2010, 8.8))` shows `Inception (2010) - 8.8/10`
- `__eq__`: two movies are equal when their **title and year** match (rating can differ)
- `__lt__`: compare by **rating**, so `sorted(movies)` goes from lowest to highest rating

```python

# Create a class called Movie
class Movie:

    # Initialize the movie with its title, year, and rating
    def __init__(self, title, year, rating):

        # Store the movie information
        self.title = title
        self.year = year
        self.rating = rating

    # TODO 5a: __str__
    # Controls how the movie looks when we print it
    def __str__(self):

        # Return the movie title, year, and rating as a string
        return f"{self.title} ({self.year}) - {self.rating}/10"

    # TODO 5b: __eq__
    # Checks if two movies are equal
    def __eq__(self, other):

        # Movies are equal if BOTH title and year match
        return self.title == other.title and self.year == other.year

    # TODO 5c: __lt__
    # Compares two movies by their ratings
    def __lt__(self, other):

        # Return True if this movie has a lower rating
        return self.rating < other.rating


# ---------------------------------
# TEST THE MOVIE CLASS
# ---------------------------------

# Create three movie objects
movie1 = Movie("Inception", 2010, 8.8)
movie2 = Movie("Titanic", 1997, 7.9)
movie3 = Movie("Inception", 2010, 8.8)

# Test __str__: Print movie information
print("Movie 1:", movie1)
print("Movie 2:", movie2)

# Test __eq__: Check if two movies are equal
print("Are movie 1 and movie 3 equal?", movie1 == movie3)

# Test __lt__: Compare movie ratings
print("Is Titanic rated lower than Inception?", movie2 < movie1)

# Sort movies by rating (lowest to highest)
movies = [movie1, movie2, movie3]

sorted_movies = sorted(movies)

print("\nMovies sorted by rating:")

for movie in sorted_movies:
    print(movie)
```

## Section 6 · Inheritance & super()
**About 4 minutes · 15 points**

A **child class** gets every attribute and method of its **parent**. It can **override** a method, and call the parent’s version with `super()`.

Run the `MenuItem` cell below (do not change it). Then write `Drink(MenuItem)`:
- `Drink(name, price, size)`: use `super().__init__(name, price)` for name and price, then store `size` (`"S"`, `"M"` or `"L"`).
- Override `label()` so it includes the size and adds the size extra from `EXTRA`:
  `Drink("Latte", 4.50, "L").label()` → `Latte (L): $5.50`

```python
class MenuItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def label(self):
        return f"{self.name}: ${self.price:.2f}"

print(MenuItem("Bagel", 3).label())
```

```python

# -----------------------------------------
# PARENT CLASS (Given by the assignment)
# -----------------------------------------

class MenuItem:

    def __init__(self, name, price):
        # Store the name of the menu item
        self.name = name

        # Store the original price
        self.price = price

    def label(self):
        # Display the name and price with 2 decimal places
        return f"{self.name}: ${self.price:.2f}"


# -----------------------------------------
# CHILD CLASS: DRINK
# -----------------------------------------

# TODO 6a: Drink inherits from MenuItem
class Drink(MenuItem):

    # Extra price depending on the drink size
    EXTRA = {"S": 0.00, "M": 0.50, "L": 1.00}

    def __init__(self, name, price, size):

        # TODO 6b: Use super() to call the parent's __init__
        # This stores the name and original price
        super().__init__(name, price)

        # TODO 6c: Store the drink size
        self.size = size

    def label(self):

        # TODO 6d: Calculate the final price
        # Add the size extra to the original price
        final_price = self.price + self.EXTRA[self.size]

        # Return the drink name, size, and final price
        # :.2f displays exactly 2 decimal places
        return f"{self.name} ({self.size}): ${final_price:.2f}"


# -----------------------------------------
# TEST THE CODE
# -----------------------------------------

# Create a small, medium, and large drink
drink1 = Drink("Latte", 4.50, "S")
drink2 = Drink("Latte", 4.50, "M")
drink3 = Drink("Latte", 4.50, "L")

# Print each drink and its adjusted price
print(drink1.label())
print(drink2.label())
print(drink3.label())
```

```python
# TODO 7: write EmailAlert and SMSAlert

# -----------------------------------------
# SECTION 7: POLYMORPHISM
# -----------------------------------------

# Create the EmailAlert class
class EmailAlert:

    # Initialize the email address
    def __init__(self, email):
        self.email = email

    # Create a send method that returns an email message
    def send(self, message):

        # Return the email address and message
        return f"Email to {self.email}: {message}"


# Create the SMSAlert class
class SMSAlert:

    # Initialize the phone number
    def __init__(self, phone):
        self.phone = phone

    # Create a send method that returns an SMS message
    def send(self, message):

        # Cut the message to its first 20 characters
        short_message = message[:20]

        # Return the phone number and shortened message
        return f"SMS to {self.phone}: {short_message}"


# -----------------------------------------
# TEST THE CODE
# -----------------------------------------

# Create an email alert
email = EmailAlert("ana@school.edu")

# Create a text message alert
sms = SMSAlert("555-0100")

# Test the email send method
print(email.send("Hi"))

# Test the SMS send method
print(sms.send("Hi"))

# Test SMS with a longer message
print(sms.send("Hello! This is a very long message."))

```

## Section 8 · Encapsulation with @property
**About 3 minutes · 10 points**

**Encapsulation:** the object protects its own rules. A `@property` looks like a normal attribute from outside, but a method checks every new value.

**Your task:** complete `Thermostat`.
- The temperature is stored in `self._temp` (the underscore means “internal”).
- `t.temp` returns it.
- Setting `t.temp = value` raises `ValueError` if `value` is below 10 or above 30.
- `__init__` must set the starting value **through the property**, so `Thermostat(5)` is rejected too.

```python
class Thermostat:
    def __init__(self, temp):
        # TODO 8a: Set the starting temperature through the property
        self.temp = temp

    @property
    def temp(self):
        # TODO 8b: Return the stored temperature
        return self._temp

    @temp.setter
    def temp(self, value):
        # TODO 8c: Make sure the temperature is between 10 and 30
        if value < 10 or value > 30:
            raise ValueError("Temperature must be between 10 and 30")

        self._temp = value
```

## Bonus · Composition (not graded)

**Composition** is a “has-a” relationship: a `Garden` *has* plants; it is not a kind of plant. It stores `Plant` objects from Section 1 in a list.

Write `Garden` with:
- `add(plant)`: adds a `Plant` object
- `thirsty()`: returns a list of the **names** of plants with `water_level` below 20
- `__len__`: so `len(garden)` is the number of plants

```python
class Garden:
    def __init__(self):
        # Start with an empty list of plants
        self.plants = []

    def add(self, plant):
        # Add a plant to the list
        self.plants.append(plant)

    def thirsty(self):
        # Return names of plants with water_level below 20
        return [plant.name for plant in self.plants if plant.water_level < 20]

    def __len__(self):
        # Return the total number of plants
        return len(self.plants)
```
