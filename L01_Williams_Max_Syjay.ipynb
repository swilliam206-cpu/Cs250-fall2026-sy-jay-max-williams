## Part 5: Debugging Practice

⏱ **Target time: 10 minutes**


📘 **Read**

**Debugging** is finding and fixing problems in a program. Every programmer does it every day.
An error message is not a scolding — it is the most useful information on your screen.

Three kinds of problems, which differ by **when** they happen:

| Type | When it happens | What you see |
|---|---|---|
| **Syntax error** | Before anything runs | Python cannot understand the structure of your code, so the program never starts. |
| **Runtime error** | Partway through | The program starts, then hits something it cannot do and stops. |
| **Logic error** | Never announced | The program runs and finishes, but the result is not what you intended. |

**How to read an error message:** start at the **last line**. It names the type of error and
gives a short description. The lines above it show where Python was when the problem occurred.
